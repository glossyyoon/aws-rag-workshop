# 3. Amazon Bedrock Agent - Connecting Knowledge base
- Tested in Amazon SageMaker AI - Notebook - JupyterLab environment.
- Kernel : conda_python3



If you connect Amazon Bedrock's Knowledge Bases (Amazon Neptune Analytics, Amazon OpenSearch Serverless) created earlier to **Amazon Bedrock Agent**, you can easily implement logic that automatically selects and searches Vector DB/Graph DB according to the user's request and outputs the results.

## 1. Prerequisites

### 1.1. Verify previously created Knowledge Base

- Log in to AWS Management Console
- Select "Amazon Bedrock" from the service list
- Select "Knowledge base" from the left panel
- Verify that Knowledge bases created in previous modules are visible




## 2. Create Amazon Bedrock Agent and Connect Knowledge Base

### 2.1. Access Amazon Bedrock Console and Create Agent

- Select "Agents" from the left navigation panel
- Click "Create agent" button



### 2.2. Configure Agent Basic Information
- Enter Agent name
- Enter Agent description (optional)
- Click "Create"




- Select default LLM model (e.g., Claude 3.5 Sonnet v2)
- Enter the following instructions in the "Instructions for the Agent" field

  - You are an assistant that utilizes both document knowledge base and knowledge graph. You search for relevant information in documents, query entity relationships in graphs, and integrate the two types of information. Graphs are specialized in understanding relationships and structures, while vector search is specialized in searching detailed content, providing answers with richer context.





- Click "Save and exit"

### 2.3. Connect Knowledge Base

- Click "Edit in Agent Builder" button




- Click "Add knowledge base" button






- Select the KB you want to connect from the list of previously created Knowledge Bases and enter instructions
- Connect both graphrag-workshop and aosrag-workshop one by one.

  - graphrag-workshop
  > <code>You are an expert in understanding complex relationships and contexts. Analyze connections between various concepts, understand interactions and causal relationships, and provide comprehensive answers. Utilize graph-based knowledge to help structural understanding of complex questions and present related information together.</code>



  - aosrag-workshop
  > <code>You are an expert in accurate information retrieval. Find the most relevant information to user questions from the knowledge base and answer clearly and concisely. Exclude unnecessary information and provide only content directly related to the question, and guide users to request specifically if additional information is needed.</code>



- After confirming that all Knowledge Bases are properly connected, click the "Save and exit" button at the top




### 2.4. Test Agent

- After completing Agent creation, move to the "Test" tab
- Test by entering Knowledge Base related questions in the test window
- Verify that the Agent searches for information from the Knowledge Base and responds
