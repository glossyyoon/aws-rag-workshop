# Data Preparation
The document we will use in this workshop is Amazon.com's 2023 performance report (SEC Form 10-Q), a quarterly financial report submitted to the U.S. Securities and Exchange Commission (SEC). This document includes Amazon's financial performance, risk factors, management discussions, and other regulatory disclosures.

The document includes the following key contents:
1. PART I. FINANCIAL INFORMATION consists of financial statements, management's discussion and analysis, risk factors covering legal and regulatory risks, stock repurchase programs and capital management strategies, and other information covering ongoing litigation and legal issues.
2. PART II: OTHER INFORMATION provides information on major lawsuits and legal disputes involving Amazon, key risk factors and uncertainties related to the business, and other disclosure information.
