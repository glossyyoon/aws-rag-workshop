# 1. Environment Setup
- For a smooth workshop experience, we will use Amazon SageMaker AI - Notebook - JupyterLab environment.
- The workshop will be conducted in the Oregon (us-west-2) region.

## 1. Creating an S3 Bucket
1. Enter S3 in the console search bar to create a bucket.
![setup0001](../img/setup-0001.jpg)

2. Enter the following string as the bucket name. You can check your account ID in the upper right corner of the console.
```
bedrock-kb-workshop-[your workshop account ID]
```
![setup0002](../img/setup-0002.jpg)

3. Click the upload button to upload files.
![setup0003](../img/setup-0003.jpg)

4. Click Add files to upload the previously downloaded files to the S3 bucket.
![setup0004](../img/setup-0004.jpg)

## 2. Bedrock Model Access Permissions and SageMaker IAM Setup
1. Go to Bedrock - Model access tab and click Enable specific models.
![setup001](../img/setup-001.jpg)

2. Select Titan Embeddings V2 and Claude 3 Sonnet and enable them.
![setup002](../img/setup-002.jpg)

3. Navigate to [Amazon SageMaker AI](https://us-east-1.console.aws.amazon.com/sagemaker/home?region=us-east-1#/landing).
4. Click on Notebooks to go to the page where you can run SageMaker Notebooks.
![setup00](../img/setup-01.jpg)

JupyterLab in SageMaker Studio is a Studio feature that manages the storage and computing resources needed to run JupyterLab applications. A JupyterLab application is a web-based integrated development environment (IDE) for notebooks, code, and data.
By default, JupyterLab applications come with SageMaker deployment images that include PyTorch, TensorFlow, Keras, NumPy, Pandas, Scikit-learn, and more.


5. Click GetStarted to go to the page for creating JupyterLab in SageMaker Studio.
![setup01](../img/setup-02.jpg)

6. Select the default Quick setup to create a SageMaker Domain.<br>
Quick setup allows you to easily configure the following settings:
* Create a new IAM role with AmazonSageMakerFullAccess policy
* Environment with public internet access and standard encryption
* Create SageMaker Studio
* Create shareable SageMaker Studio Notebooks
* Create SageMaker Canvas
* IAM Authentication

![setup03](../img/setup-03.jpg)

7. This process will complete in a few minutes. After creation, check Status: Ready to confirm successful creation.
![setup04](../img/setup-04.jpg)

8. Click on the domain you just created in the Domain tab.
![setup002](../img/setup-003.jpg)

9. Copy the SageMaker ExecutionRole running in the Domain.
![setup004](../img/setup-004.jpg)

10. Go to IAM - Role and search for the role you just copied.
![setup005](../img/setup-005.jpg)

11. Click on the customer managed policy among the Permissions policies of that role.
![setup006](../img/setup-006.jpg)

12. Click Edit to modify the policy.
![setup007](../img/setup-007.jpg)

13. Paste the following code and click Next - Save changes to update the role.
```json
{
    "Version": "2012-10-17",
    "Statement": [
        {
            "Action": [
                "s3:ListBucket"
            ],
            "Effect": "Allow",
            "Resource": [
                "arn:aws:s3:::SageMaker"
            ]
        },
        {
            "Action": [
            "s3:GetObject",
                "s3:PutObject",
                "s3:DeleteObject"
            ],
            "Effect": "Allow",
            "Resource": [
                "arn:aws:s3:::*"
            ]
        },
        {
            "Effect": "Allow",
            "Action": [
                "bedrock:*"
            ],
            "Resource": "*"
        }
    ]
}

```


14. Review the code and click Next to save.
![setup008](../img/setup-008.jpg)

## 3. Running SageMaker Studio

1. Go to the Studio tab, make sure you've selected the domain with the same name as the one you just created, and click Open Studio to launch Studio.
![setup05](../img/setup-05.jpg)

2. Access SageMaker Studio and run JupyterLab applications.
![setup06](../img/setup-06.jpg)

3. Click Create JupyterLab space.
![setup07](../img/setup-07.jpg)

4. Enter the following as the JupyterLab name and click Create space.
```code{language=md showLineNumbers=false showCopyAction=true}
advanced-rag-workshop
```
![setup08](../img/setup-08.jpg)

5. Once the JupyterLab is created, click Run space to run it.
![setup09](../img/setup-09.jpg)

6. Click Open JupyterLab to launch the application.
![setup10](../img/setup-10.jpg)

All preparations are now complete.
![setup10](../img/setup-11.jpg)
